## 1. Imports and setup
The following cell imports the libraries needed for this exercise.

```python
import numpy as np, pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, roc_auc_score
rng=np.random.default_rng(10); n=600
age=rng.integers(18,65,n); income=rng.normal(55000,18000,n).clip(15000,120000)
logit=-5 + .035*age + income/25000
prob=1/(1+np.exp(-logit)); bought=rng.binomial(1,prob)
df=pd.DataFrame({'age':age,'income':income,'bought':bought}); X=df[['age','income']]; y=df.bought
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,random_state=42,stratify=y)
model=make_pipeline(StandardScaler(),LogisticRegression()).fit(Xtr,ytr); pred=model.predict(Xte); p=model.predict_proba(Xte)[:,1]
print('Accuracy:',round(accuracy_score(yte,pred),3),'ROC-AUC:',round(roc_auc_score(yte,p),3)); print(confusion_matrix(yte,pred))
print('Purchase probability for age 30, income 50000:',round(model.predict_proba(pd.DataFrame([[30,50000]],columns=X.columns))[0,1],3))
```
