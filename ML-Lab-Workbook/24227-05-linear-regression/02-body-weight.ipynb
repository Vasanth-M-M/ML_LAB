## 1. Imports and setup
The following cell imports the libraries needed for this exercise.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import r2_score, mean_squared_error
rng=np.random.default_rng(7); n=500
height=rng.normal(1.68,.1,n).clip(1.4,2.0); age=rng.integers(18,70,n); male=rng.integers(0,2,n)
weight=51+8.6*height-.12*age+22.9*male+rng.normal(0,15,n)
df=pd.DataFrame({'Height':height,'Age':age,'Gender_Male':male,'Weight':weight})
X=df[['Height','Age','Gender_Male']]; y=df['Weight']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42)
model=LinearRegression().fit(X_train,y_train); pred=model.predict(X_test)
print('Intercept:',round(model.intercept_,3)); print(pd.Series(model.coef_,index=X.columns))
print('Test R2:',round(r2_score(y_test,pred),3),'RMSE:',round(mean_squared_error(y_test,pred)**.5,2))
print('5-fold CV R2:',cross_val_score(model,X,y,cv=5,scoring='r2').mean().round(3))
print('Predicted weight:',round(model.predict(pd.DataFrame([[1.75,30,1]],columns=X.columns))[0],1),'kg')
plt.scatter(y_test,pred,alpha=.6); plt.xlabel('Actual weight'); plt.ylabel('Predicted weight'); plt.show()
```
