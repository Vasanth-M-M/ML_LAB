## 1. Imports and setup
The following cell imports the libraries needed for this exercise.

```python
import pandas as pd, matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
# Synthetic two-cluster data mirrors the classic short/long eruption pattern.
X,_=make_blobs(n_samples=272,centers=[[-1.2,-1.0],[1.3,1.1]],cluster_std=[.45,.55],random_state=42)
df=pd.DataFrame(X,columns=['waiting','duration']); scaler=StandardScaler(); Xs=scaler.fit_transform(df)
wcss=[]; sil={}
for k in range(1,9):
 km=KMeans(n_clusters=k,n_init=10,random_state=42).fit(Xs); wcss.append(km.inertia_)
 if k>1: sil[k]=silhouette_score(Xs,km.labels_)
fig,ax=plt.subplots(1,2,figsize=(10,4)); ax[0].plot(range(1,9),wcss,'o-'); ax[0].set(xlabel='K',ylabel='WCSS',title='Elbow method'); ax[1].bar(sil.keys(),sil.values()); ax[1].set(xlabel='K',ylabel='Silhouette',title='Silhouette score'); plt.tight_layout(); plt.show()
model=KMeans(n_clusters=2,n_init=10,random_state=42).fit(Xs); df['cluster']=model.labels_; print(df.groupby('cluster').mean().round(2)); print('Silhouette:',round(silhouette_score(Xs,model.labels_),3))
plt.scatter(df.waiting,df.duration,c=df.cluster); plt.xlabel('Waiting (scaled demo units)'); plt.ylabel('Duration (scaled demo units)'); plt.show()
```
