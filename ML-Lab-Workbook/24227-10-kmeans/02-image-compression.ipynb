## 1. Imports and setup
The following cell imports the libraries needed for this exercise.

```python
import numpy as np, matplotlib.pyplot as plt, math
from sklearn.datasets import load_sample_image
from sklearn.cluster import KMeans
image=load_sample_image('china.jpg'); img=image[10:406,120:516]
pixels=img.reshape(-1,3)/255.0; h,w,_=img.shape; compressed={}; rows=[]
for k in [2,4,8,16,32]:
 km=KMeans(n_clusters=k,n_init=1,random_state=42).fit(pixels); recon=km.cluster_centers_[km.labels_].reshape(h,w,3); compressed[k]=recon
 mse=np.mean((img/255.0-recon)**2); psnr=10*np.log10(1/mse); ratio=(h*w*24)/(h*w*math.ceil(math.log2(k))+k*24)
 rows.append({'K colours':k,'bits/pixel':math.ceil(math.log2(k)),'PSNR (dB)':psnr,'Compression ratio':ratio})
import pandas as pd
print(pd.DataFrame(rows).round(2))
fig,axes=plt.subplots(2,3,figsize=(12,8)); axes[0,0].imshow(img); axes[0,0].set_title('Original')
for ax,(k,im) in zip(axes.ravel()[1:],compressed.items()): ax.imshow(im); ax.set_title(f'K={k} colours')
for ax in axes.ravel(): ax.axis('off')
plt.tight_layout(); plt.show()
```
