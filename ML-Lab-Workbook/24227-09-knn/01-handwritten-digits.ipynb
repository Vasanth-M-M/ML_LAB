## 1. Imports and setup
The following cell imports the libraries needed for this exercise.

```python
import numpy as np, matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report
D=load_digits(); X,y=D.data,D.target
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,random_state=42,stratify=y)
results=[]
for k in range(1,16):
 m=KNeighborsClassifier(n_neighbors=k).fit(Xtr,ytr); results.append((k,m.score(Xte,yte)))
best_k=max(results,key=lambda z:z[1])[0]; model=KNeighborsClassifier(n_neighbors=best_k).fit(Xtr,ytr); pred=model.predict(Xte)
print('Best K:',best_k,'Accuracy:',round(accuracy_score(yte,pred),3)); print(classification_report(yte,pred))
plt.plot([r[0] for r in results],[r[1] for r in results],marker='o'); plt.xlabel('K'); plt.ylabel('Test accuracy'); plt.show()
fig,axes=plt.subplots(2,5,figsize=(9,4))
for ax,img,label in zip(axes.ravel(),D.images[:10],D.target[:10]): ax.imshow(img,cmap='gray_r'); ax.set_title(str(label)); ax.axis('off')
plt.tight_layout(); plt.show()
```
