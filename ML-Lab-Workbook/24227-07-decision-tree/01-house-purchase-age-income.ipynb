## 1. Imports and setup
The following cell imports the libraries needed for this exercise.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
from sklearn.metrics import accuracy_score, classification_report
housing=fetch_california_housing(as_frame=True)
df=housing.frame[['HouseAge','MedInc','MedHouseVal']].rename(columns={'HouseAge':'age','MedInc':'income','MedHouseVal':'house_value'})
df['bought_house']=(df.house_value>df.house_value.median()).astype(int)
X=df[['age','income']]; y=df.bought_house
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,random_state=42,stratify=y)
model=DecisionTreeClassifier(max_depth=3,random_state=42).fit(Xtr,ytr)
print('Accuracy:',round(accuracy_score(yte,model.predict(Xte)),3)); print(export_text(model,feature_names=list(X.columns)))
plt.figure(figsize=(14,7)); plot_tree(model,feature_names=X.columns,class_names=['Below median','Above median'],filled=True); plt.show()
```
