## 1. Imports and setup
The following cell imports the libraries needed for this exercise.

```python
import pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
from sklearn.metrics import accuracy_score
weather=pd.DataFrame([
['Sunny','Hot','High','Weak','Stay in'],['Sunny','Hot','High','Strong','Stay in'],['Overcast','Hot','High','Weak','Go out'],['Rain','Mild','High','Weak','Go out'],['Rain','Cool','Normal','Weak','Go out'],['Rain','Cool','Normal','Strong','Stay in'],['Overcast','Cool','Normal','Strong','Go out'],['Sunny','Mild','High','Weak','Stay in'],['Sunny','Cool','Normal','Weak','Go out'],['Rain','Mild','Normal','Weak','Go out'],['Sunny','Mild','Normal','Strong','Go out'],['Overcast','Mild','High','Strong','Go out'],['Overcast','Hot','Normal','Weak','Go out'],['Rain','Mild','High','Strong','Stay in']],columns=['Outlook','Temperature','Humidity','Wind','Decision'])
X=pd.get_dummies(weather.drop(columns='Decision')).astype(int); y=weather.Decision
model=DecisionTreeClassifier(criterion='entropy',random_state=42).fit(X,y)
print('Training accuracy:',round(accuracy_score(y,model.predict(X)),3)); print(export_text(model,feature_names=list(X.columns)))
plt.figure(figsize=(14,7)); plot_tree(model,feature_names=X.columns,class_names=model.classes_,filled=True); plt.show()
```
