## 1. Imports and setup
The following cell imports the libraries needed for this exercise.

```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.pipeline import make_pipeline
from sklearn.metrics import classification_report, accuracy_score
# Two text categories serve as a runnable text-classification demonstration.
data=fetch_20newsgroups(subset='all',categories=['rec.autos','sci.med'],remove=('headers','footers','quotes'))
Xtr,Xte,ytr,yte=train_test_split(data.data,data.target,test_size=.25,random_state=42,stratify=data.target)
model=make_pipeline(TfidfVectorizer(max_features=10000),SVC(kernel='linear')).fit(Xtr,ytr); pred=model.predict(Xte)
print('Accuracy:',round(accuracy_score(yte,pred),3)); print(classification_report(yte,pred,target_names=data.target_names))
print('Note: replace the two-category demonstration with the SMS Spam Collection dataset for the exact spam/not-spam task.')
```
