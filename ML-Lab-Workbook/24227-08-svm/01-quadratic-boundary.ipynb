## 1. Imports and setup
The following cell imports the libraries needed for this exercise.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
rng=np.random.default_rng(0); n=500
x1=rng.uniform(-3,3,n); x2=rng.uniform(-2,10,n); y=(x2>x1**2+rng.normal(0,.5,n)).astype(int)
X=np.c_[x1,x2]; Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,random_state=42,stratify=y)
models={'Linear':SVC(kernel='linear'),'Polynomial (degree 2)':SVC(kernel='poly',degree=2,C=10,coef0=1),'RBF':SVC(kernel='rbf',C=10)}
rows=[]
for name,m in models.items():
 m.fit(Xtr,ytr); rows.append({'Kernel':name,'Train accuracy':m.score(Xtr,ytr),'Test accuracy':m.score(Xte,yte),'Support vectors':m.n_support_.sum()})
print(pd.DataFrame(rows).round(3))
fig,axes=plt.subplots(1,3,figsize=(15,4)); xx,yy=np.meshgrid(np.linspace(-3.2,3.2,200),np.linspace(-2.5,10.5,200))
for ax,(name,m) in zip(axes,models.items()):
 z=m.predict(np.c_[xx.ravel(),yy.ravel()]).reshape(xx.shape); ax.contourf(xx,yy,z,alpha=.25,cmap='coolwarm'); ax.scatter(Xte[:,0],Xte[:,1],c=yte,s=10,cmap='coolwarm'); ax.set_title(f'{name}: {m.score(Xte,yte):.2f}')
plt.tight_layout(); plt.show()
```
