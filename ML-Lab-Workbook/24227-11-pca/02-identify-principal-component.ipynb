## 1. Imports and setup
The following cell imports the libraries needed for this exercise.

```python
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
load=pd.DataFrame({'PC1':[.190,.544,.782,.365,.585,.394,.985,.520,.142], 'PC2':[.017,.020,-.605,.294,.085,-.273,.126,.402,.150], 'PC3':[.207,.204,.144,.585,.234,.027,-.111,.519,.239]},index=['Climate','Housing','Health','Crime','Transportation','Education','Arts','Recreation','Economy'])
load.index.name='Variable'
print('Dominant variable in each component:')
for pc in load.columns:
 v=load[pc].abs().idxmax(); print(f'{pc}: {v}, loading={load.loc[v,pc]:+.3f}')
print('Variables with |loading| >= 0.5:')
for pc in load.columns: print(pc,load.index[load[pc].abs()>=.5].tolist())
sns.heatmap(load,annot=True,fmt='.3f',cmap='RdBu_r',center=0,vmin=-1,vmax=1); plt.title('Component loadings'); plt.show()
print('Overall importance across PC1–PC3:'); print((load**2).sum(axis=1).sort_values(ascending=False).round(3))
```
